```python
#step1: Init llm
!pip install langchain-cohere
from google.colab import userdata
LLM_KEY = userdata.get("COHERE_API_KEY")
from langchain_cohere import ChatCohere
llm=ChatCohere(
    cohere_api_key=LLM_KEY,
    model="command-a-03-2025"
)
```

```python
#step2:tool/function(calculator so that LLM can do complex Multiplications)
def tool_1(a,b):
  """Multiplies tool which performs the product of two interger numbers."""
  return a*b
```

```python
#step3:bind/connect llm with tools into a ReAct Architecure (Agent)
from langchain.agents import create_agent
agent=create_agent(
    model=llm,
    tools=[tool_1]
)
```

```python
#step 4: call the agent on user quiries

from langchain_core.tools import tool
from langchain.agents import create_agent

# Redefine tool_1 with @tool decorator for explicit schema generation
@tool
def tool_1(a: int, b: int) -> int:
  """Multiplies two numbers. Useful for when you need to perform multiplication.

  Args:
    a: The first integer number.
    b: The second integer number.
  """
  return a*b

# Recreate the agent with the correctly defined tool_1
# Assuming llm is already defined from a previous cell
agent=create_agent(
    model=llm,
    tools=[tool_1]
)

response=agent.invoke({"messages":"hi"})
for msg in response["messages"]:
  msg.pretty_print()
```

```python

response=agent.invoke({"messages":"what is 123456*789012"})
for msg in response["messages"]:
  msg.pretty_print()
```
